## FILE OPERATOR📌

```python
import os
from datetime import datetime


class JournalManager:
    def __init__(self, filename="journal.txt"):
        self.filename = filename

    # Add new entry
    def add_entry(self):
        try:
            entry = input("Write your journal entry: ")

            timestamp = datetime.now().strftime("%d-%m-%Y %H:%M:%S")

            with open(self.filename, "a") as file:
                file.write(f"\n[{timestamp}] {entry}\n")

            print("Entry added successfully!")

        except Exception as e:
            print("Error while saving entry:", e)

    # View all entries
    def view_entries(self):
        try:
            with open(self.filename, "r") as file:
                data = file.read()

                if data.strip() == "":
                    print("No journal entries found.")
                else:
                    print("\n----- JOURNAL ENTRIES -----")
                    print(data)

        except FileNotFoundError:
            print("Journal file does not exist yet.")

        except PermissionError:
            print("Permission denied.")

    # Search entry
    def search_entry(self):
        try:
            keyword = input("Enter keyword to search: ")

            with open(self.filename, "r") as file:
                lines = file.readlines()

            found = False

            print("\nSearch Results:")

            for line in lines:
                if keyword.lower() in line.lower():
                    print(line.strip())
                    found = True

            if not found:
                print("No matching entry found.")

        except FileNotFoundError:
            print("Journal file not found.")

    # Delete all entries
    def delete_entries(self):
        try:
            choice = input(
                "Are you sure you want to delete all entries? (yes/no): "
            )

            if choice.lower() == "yes":

                if os.path.exists(self.filename):
                    os.remove(self.filename)
                    print("All journal entries deleted.")

                else:
                    print("Journal file does not exist.")

            else:
                print("Deletion cancelled.")

        except Exception as e:
            print("Error:", e)


# Main Program
def main():

    journal = JournalManager()

    while True:

        print("\n====== PERSONAL JOURNAL ======")
        print("1. Add New Entry")
        print("2. View All Entries")
        print("3. Search Entry")
        print("4. Delete All Entries")
        print("5. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            journal.add_entry()

        elif choice == "2":
            journal.view_entries()

        elif choice == "3":
            journal.search_entry()

        elif choice == "4":
            journal.delete_entries()

        elif choice == "5":
            print("Thank you for using Journal Manager!")
            break

        else:
            print("Invalid choice. Please try again.")


main()
```
